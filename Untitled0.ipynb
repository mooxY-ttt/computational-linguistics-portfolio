```python
# 📚 Stylometric Analysis of Classic Literature: A Computational Approach

## 🌟 Motivation
As an aspiring Computational Linguistics student, I initiated this micro-project to explore how Python and foundational Natural Language Processing (NLP) techniques can be utilized to quantitatively analyze and compare the stylistic features of different authors. This project serves as my first step in bridging linguistic theory with computational methods.

## 🛠️ Tools Used
- **Programming Language**: Python 3
- **Core Libraries**:
  - `re` (Regular Expressions for text normalization and cleaning)
  - `collections.Counter` (Frequency analysis)
  - `matplotlib` (Data visualization)
- **Development Environment**: Google Colab

## 📖 Data Source
- *Alice's Adventures in Wonderland* by Lewis Carroll
- *The Hound of the Baskervilles* by Arthur Conan Doyle
- Both texts were programmatically fetched and extracted from the **Project Gutenberg** open-source library.

## 💡 Linguistic Insights
The quantitative visualization reveals distinct stylistic patterns between the two authors:
1. **Conan Doyle** employs a higher relative frequency of concrete nouns and action-oriented verbs (e.g., "man", "door", "said") to drive the suspenseful, plot-heavy narrative of detective fiction.
2. **Carroll**, while also heavily utilizing "said" due to the dialogue-rich nature of the text, frequently pairs it with character-specific or fantastical vocabulary (e.g., "queen", "mock").
This preliminary analysis demonstrates the potential of frequency-based methods in **Stylometry** (computational stylistics) to objectively validate literary observations.

## ⚠️ Limitations & Future Work
Currently, the text tokenization relies on basic Regular Expressions, which may struggle with complex contractions (e.g., "don't") or morphological variations. As my next step, I plan to integrate advanced NLP libraries like `spaCy` or `NLTK` to implement precise **Part-of-Speech (POS) tagging** and **Lemmatization**, allowing for a more granular syntactic analysis.

## 🚀 How to Run
This project is entirely cloud-based. You can open and execute all the code directly in Google Colab by clicking the badge below:

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/)
*(Note: Please replace the link above with your actual shared Colab notebook link)*
```

```python
# Import necessary libraries for data fetching, text processing, and visualization
import urllib.request       # For downloading text files from the web
import re                   # Regular expressions for text cleaning
from collections import Counter # For counting word frequencies efficiently
import matplotlib.pyplot as plt # For data visualization

print("All libraries imported successfully!")
```

```python
# URLs for the public domain texts from Project Gutenberg
url_alice = "https://www.gutenberg.org/cache/epub/11/pg11.txt"
url_sherlock = "https://www.gutenberg.org/cache/epub/2852/pg2852.txt"

# Download and decode the text files
print("⏳ Downloading 'Alice in Wonderland'...")
response1 = urllib.request.urlopen(url_alice)
text_alice = response1.read().decode('utf-8')

print("⏳ Downloading 'The Hound of the Baskervilles'...")
response2 = urllib.request.urlopen(url_sherlock)
text_sherlock = response2.read().decode('utf-8')

print(f" Download complete! Alice length: {len(text_alice)} chars, Sherlock length: {len(text_sherlock)} chars")
```

```python
def clean_and_tokenize(text):
    # Step 1: Convert all text to lowercase for consistency
    text = text.lower()

    # Step 2: Use Regular Expressions to remove all non-alphabetic characters (punctuation, numbers, newlines)
    # Replace them with a space to prevent words from merging
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Step 3: Tokenize the text by splitting it into a list of individual words
    words = text.split()

    return words

# Apply the function to both texts
words_alice = clean_and_tokenize(text_alice)
words_sherlock = clean_and_tokenize(text_sherlock)

print(f"✅ Tokenization complete! Alice words: {len(words_alice)}, Sherlock words: {len(words_sherlock)}")
```

```python
# Define a set of common English stop words to filter out (they carry little semantic meaning)
stop_words = set(['the', 'and', 'a', 'to', 'of', 'in', 'i', 'is', 'that', 'it', 'on', 'you', 'this',
                  'for', 'but', 'with', 'are', 'have', 'be', 'at', 'or', 'as', 'was', 'so', 'if',
                  'out', 'not', 'an', 'all', 'he', 'she', 'they', 'we', 'his', 'her', 'their'])

def get_top_words(word_list, stop_words_set, top_n=10):
    # Filter out stop words and single-character words (like 's' or 't' from contractions)
    filtered_words = [w for w in word_list if w not in stop_words_set and len(w) > 1]

    # Count the frequency of each remaining word
    word_counts = Counter(filtered_words)

    # Return the top N most common words
    return word_counts.most_common(top_n)

# Get the top 10 words for each book
top_alice = get_top_words(words_alice, stop_words)
top_sherlock = get_top_words(words_sherlock, stop_words)

print(" Top 10 words in Alice in Wonderland:", top_alice)
print(" Top 10 words in The Hound of the Baskervilles:", top_sherlock)
```

```python
# Extract labels (words) and counts (frequencies) for plotting
labels_alice = [item[0] for item in top_alice]
counts_alice = [item[1] for item in top_alice]

labels_sherlock = [item[0] for item in top_sherlock]
counts_sherlock = [item[1] for item in top_sherlock]

# Set up the figure size for better readability
plt.figure(figsize=(12, 6))

# Define x-axis positions
x = range(len(labels_alice))

# Plot bar charts for both books with slight offset for side-by-side comparison
plt.bar(x, counts_alice, width=0.4, color='skyblue', label="Alice in Wonderland", align='center')
plt.bar([i + 0.4 for i in x], counts_sherlock, width=0.4, color='lightcoral', label="The Hound of the Baskervilles", align='center')

# Add titles and labels
plt.xlabel('Most Frequent Content Words', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Stylometric Comparison: Word Frequency Distribution', fontsize=14, fontweight='bold')

# Set x-axis ticks to show the words
plt.xticks([i + 0.2 for i in x], labels_alice, rotation=45, ha='right')
plt.legend()

# Adjust layout to prevent labels from being cut off
plt.tight_layout()

# Display the plot
plt.show()
```
